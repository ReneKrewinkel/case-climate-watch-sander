```python
import pandas as pd 
from sqlalchemy import create_engine
engine = create_engine("mysql+pymysql://root@localhost:3306/climate_watch")

with engine.connect() as con:
    print("succesful connection")
```

```python
query = """
select * from globaltemperatures
"""
df = pd.read_sql(query, engine)
display(df.head())
```

```python
df['dt'] = pd.to_datetime(df['dt'])
df_subset_globtemp = df[df['dt'] > '1979-12-31']
display(df_subset_globtemp.head())
df_subset_globtemp.to_csv("C:/Users/sande/Downloads/subset_globtemp.csv", index=False)
```
